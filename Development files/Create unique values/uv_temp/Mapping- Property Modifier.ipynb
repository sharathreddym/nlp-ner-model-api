```python
def add_at(text, split_by=","):
    if split_by=="(":
        text_splits = text.split("(")
        text_splits = [i.strip() for i in text_splits]
        use_at = 0
        
        for i in range(1, len(text_splits)):
            if text_splits and (text_splits[i][:1].isdigit() or text_splits[i][:1] in ['+', '-']):
                use_at = i
                break

        if use_at:
#             print("at " + prop_name[use_at])
            text_splits[use_at] = "at " + text_splits[use_at]
            text = "(".join(text_splits)
            
    else:
        text_splits = text.split(",")
        text_splits = [i.strip() for i in text_splits]
        use_at = 0
        for i in range(1, len(text_splits)):
            if text_splits and (text_splits[i][:1].isdigit() or text_splits[i][:1] in ['+', '-']):
                use_at = i
                break

        if use_at:
    #         print("at " + text_splits[use_at-1])
            text_splits[use_at] = "at " + text_splits[use_at]
            text = ", ".join(text_splits)   

    return text
```

```python
property_details = {}
remove = ["-", "--"]
for property_name, group in df1.groupby('PROPERTY_NAME'):
    property_name = property_name.strip()
    property_names = [property_name]
    split = False
    
    if ',' in property_name: 
        property_name = add_at(property_name, ",")
    elif '(' in property_name:
        property_name = add_at(property_name, "(")
        
    if ',' in property_name:
        property_name = property_name.split(",")
        for i in range(1, len(property_name)+1):
            prop_names = property_name[0: i]
            if len(prop_names) > 1:
                prop_name = " ".join(prop_names)
#                 prop_name2 = ",".join(prop_name)
#                 property_names.extend([prop_name1, prop_name2])
                property_names.append(prop_name)
            else:
                property_names.append(prop_names[0])
            
        split = True
        #"Tear strength, Method Ba, perpendicular" >>>>>>>>>>>"Tear strength"
    else:
        pass
        
   
    #group['modifier'] = group["VALUE_ASSMNT_SI"] + " " + group["UNIT_OF_MEAS_SI"]
    group['modifier'] = group["UNIT_OF_MEAS_SI"]
    modifiers = list(group['modifier'].dropna().unique())
    modifiers = [modifier.strip() for modifier in modifiers]
    modifiers = [i for i in modifiers if i]
    modifiers = [i for i in modifiers if i not in remove]
    
    property_names = list(set(property_names))
#     if len(property_names) > 1:
#         print(property_names, "\n")
    for p in property_names:
        if "(" in p and ")" not in p:
            p = p.replace("(","")
        
        if '@ ' in p:
            p = p.replace('@',random.choice(['at','@']))
        else:
            pass
        
        if p in property_details:
            property_details[p] = list(set(property_details[p] + modifiers))
        else:
            property_details[p] = modifiers
            
# property_details
```
