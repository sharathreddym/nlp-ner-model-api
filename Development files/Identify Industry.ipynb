```python
import re
def get_industry(application):
    """
    This function takes an application and returns the corresponding industry
    from a predefined list, or None if no industry is found.

    Args:
      application: A string describing the application.

    Returns:
      A string representing the industry or None if not found.
    """
    #separate logic for Electrical & Electronics
    if "electrical" in  application and "electronic" in application:
        return "Electrical & Electronics"
    
    stop_words = ["an", "and", "are", "as", "at", "be", "but", "by", "for", "if", "in", "into", "is",
                    "it", "no", "not", "of", "on", "or", "such", "that", "the", "their", "then", "there",
                    "these", "they", "this", "to", "was", "will", "with"]
    app_keywords = [i for i in re.split(r'[ -/&]+', application) if i not in stop_words]
    
    industries = {
        "Medical & Pharma": ["medical", "pharma","medicine","drug", "hospital", "health", "healthcare","pharmaceutical", "clinic", "clinical", "biotechnology", "biomedical", "dental", "surgical"],

        "Industrial": ["industrial", "factory", "manufacturing","engineering","machinery"],

        "Automotive & Transportation": ["automotive", "automobile", "automobility","car", "cars", "bike","motorbike","vehicle", "truck","transportation", "transport", "mobility", "railway", "aviation", "aerospace"],

        "Consumer Goods": ["consumer", "household", "apparel", "retail", "furniture", "footwear", "appliance"] 
    }
    
    for app_keyword in app_keywords:
        matching_industries = [industry for industry, keywords in industries.items() if any(keyword in app_keyword and len(keyword)/len(app_keyword)>=0.8 for keyword in keywords)]  
        if matching_industries:
            return matching_industries[0]
    
    return None
```

```python

import json
unique_values = json.load(open("../dependencies/unique_values_22_02_24.json"))
```

```python
all_applications = unique_values['APPLICATION']
```

```python
predictions = []
for application in all_applications:
    predicted_industry = get_industry(application)
    if predicted_industry:
        predictions.append((application,predicted_industry))
        print(f"Input: '{application}' - Output: '{predicted_industry}'")
```

```python
import pandas as pd
predictions = pd.DataFrame(predictions)
# predictions.to_csv("industry predictions.csv",index=False)
```
