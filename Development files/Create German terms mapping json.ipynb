```python
import pandas as pd

import string
special_chars=string.punctuation
special_chars=special_chars.replace(')','')
special_chars=special_chars.replace(']','')
```

```python
%%time
def remove_special_characters(input_str):
    # Remove the special characters from the start & end of the string
    cleaned_str = input_str.lower().strip(special_chars)
    return cleaned_str

german_terms_df =pd.read_excel(r".\.\Data\Translations_SPT_Feature.xlsx")
german_terms_df = german_terms_df[['English','EN_short','German','DE_short']].drop_duplicates()
german_terms_df= german_terms_df.dropna().reset_index(drop=True)

german_terms_df['English'] = german_terms_df['English'].apply(lambda x: remove_special_characters(x))
german_terms_df['German'] = german_terms_df['German'].apply(lambda x: remove_special_characters(x))

german_terms_df['DE_short'] = german_terms_df['DE_short'].apply(lambda x: remove_special_characters(x))
german_terms_df['EN_short'] = german_terms_df['EN_short'].apply(lambda x: remove_special_characters(x))

german_mapping = dict(zip(german_terms_df['German'], german_terms_df['English']))
german_mapping.update(dict(zip(german_terms_df['DE_short'], german_terms_df['EN_short'])))
german_mapping.update({"spritzgießen":"injection molding",
                        "massetemperatur":"melt temperature",
                        "anisotropieverhältnis":"anisotropy ratio",
                         "werkzeugtemperatur":"mold temperature",
                         "spritzgeschwindigkeit":"injection velocity",
                         "nachdruck":"pressure at hold",
                         "temp. der plastifizierten masse":"injection temperature",
                         "nachdruck":"hold pressure",
                         "härtungszeit":"cure time",
                         "zug-modul":"tensile modulus",
                         "zugfestigkeit":"tensile strength",
                       "temperatur":"temperature",
                       "intrinsische viskosität":"intrinsic viscosity",
                       "wasseraufnahme":"water absorption",
                       "schmelzindex":"melt index",
                       "molekulargewicht":"molecular weight",
                       "mittleres":"average",
                    "spannung":"tensile stress",
                       "biegedehnung":"flexural stress",
                       "durchstoßversuch":"puncture test",
                       "schädigungsdeformation":"deformation to damage",
                       "gesamtarbeit":"total work",
                       "schädigungsarbeit":"work to damage",
                       "maximale kraft":"force max",
                       "druckfestigkeit":"compressive stress",
                       "rockwell härte":"rockwell hardness",
                       "schweißlinie":"weldline tensile",
                       
                      })
```

```python
import json
with open("../dependencies/german_terms_mapping.json", "w") as fp:
    json.dump(german_mapping , fp)
```
