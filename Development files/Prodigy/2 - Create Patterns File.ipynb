```python
import json
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("gel spinning (fibers)")
[{"LOWER": t.lower_} for t in doc]
```

```python
def create_label(column: str):
    return {'label': column}

def create_match_patterns(unique_values: dict, path: str):
    match_patterns = []
    for label, values in unique_values.items():
    #     print(label)  
        if values:
            for val in values:
                match_pattern = create_label(label)
                pattern = []
    #             print(label, val)
    #             for word in val.split():
    #                 pattern.append({'lower': word})
                doc = nlp(val)
                match_pattern['pattern'] = [{"LOWER": t.lower_} for t in doc if not t.lower_.isspace()]
                if match_pattern['pattern']:    
                    match_patterns.append(match_pattern)

    #             with open(path, "a", encoding='utf-8') as myfile:
    #                 if(os.path.getsize(path) == 0):
    #                     myfile.write(str(match_pattern))
    #                 else:
    #                     myfile.write("\n"+ str(match_pattern))
    #             with open(path, "a", encoding="utf-8") as jsonfile:  
    #                 jsonfile.write(json.dumps(match_pattern))
    #                 jsonfile.write("\n")

        else:
            print(f"No values values in {label}")
    
    with open(path, "w", encoding="utf-8") as jsonfile:
        jsonfile.write("\n".join([json.dumps(i) for i in match_patterns]))
        
#     print("\n".join([json.dumps(i) for i in match_patterns]))
```
