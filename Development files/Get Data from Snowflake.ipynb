```python
import ast
import pymssql
import datetime
import itertools
import numpy as np
import pandas as pd
import sqlalchemy as sa
from datetime import date
from datetime import datetime as dt
from sqlalchemy import create_engine
from dateutil.parser import parse
```

```python
def transform(cert):
    x = []
    if cert == 'wras 23':
        x = ['wras 23C']
    elif cert == 'wras 60':
        x = ['wras 60C']
    elif cert == 'wras 85':
        x = ['wras 85C']         
    elif cert == 'nsf61 23C':
        x = ['nsf61 23C', 'nsf61 73F']
    elif cert == 'nsf61 82C':
        x = ['nsf61 82C', 'nsf61 179.6F']       
    elif cert == 'acs 23':
        x = ['acs 23C', 'acs 73F']        
    elif cert == 'acs 60':
        x = ['acs 60C', 'acs 140F']       
    elif cert == 'acs 85':
        x = ['acs 85C', 'acs 185F']        
    elif cert == 'ktw bwgl en16421 P2 23C':
        x = ['ktw bwgl at 23C']       
    elif cert == 'ktw bwgl en16421 P2 60C':
        x = ['ktw bwgl at 60C']        
    elif cert == 'ktw bwgl en16421 P2 85C':
        x = ['ktw bwgl at 85C']
    elif cert == 'ktw bwgl en16421 P2 en16422':
        x = ['en16421']
    else:
        x = []
    return x

def is_date(string, fuzzy=False):
    try:
        parse(string, fuzzy=fuzzy)
        return True
    except ValueError:
        return False

#function to extract year and month from datetime date
def extract_year_month(df):
    cols=['wras_23c','wras_60c','wras_85c','ktw_bwgl_23c','ktw_bwgl_60c','ktw_bwgl_85c','en16421','acs_23c','acs_60c','acs_85c','Today_date']
    for i in cols:
        i=str(i)
        df[i+'_year']=pd.to_datetime(df[i]).dt.strftime('%Y')
        df[i+'_month']=pd.to_datetime(df[i]).dt.strftime('%m')
    return df

char1 = [
    'XXXXXX', 
    '..Other none', 
    '..Other', 'none', 
    '\n\t\t', 
    'as per SPAL Automotive', 
    '/ 1. Page material data', 
    'listed', 
    '"', "'", 
    'table 2', 
    'Only - not officially approved but best fit from Celanese portfolio to spec requirements',
    '(application specific)',
    'Color approved',
    '(limited approval)',
    'Limited approval for ',
    '(grey coloured masterbatch)',
    '(blue coloured masterbatch)',
    '100% COLOR MATCH',
    '100% Color Match',
    '(pre-compounded or Salt; Pepper)',
    '(Nanjing)',
    'Validated for Strut sensor by HGT',
    'Exterior',
    'Interior',
    'China',
    'as ',
    '*',
    'Specified on each application ',
    '(Campo Bom)',
    '(COLORS)',
    '(Kalahari)',
]

char2 = [
    '&amp',
    'no spec',
    '\xa0',
    '\t\t\t\t\t\t\t\t\t\t\t',
    '\t\t\t\t\t\t\t\t\t\t\t\t\t\t\t\t\t\t',
    '<br />\t',
    '\t\t\t',
    '\t\t',
    '</th>',
    '<br />\n',
    '<br /><a href=https://iq.ul.com/ul/cert.aspx?ULID=102512443>',
    '</a>',
    '\n \n\n<tr>',
    ' \t<th>',
    '<th>',
    '<tr>',
    '- attached email/',
    'FTM Material Data Sheet PSA - attached',
    'No code - APTIV MSPEC document',
    'E ;E',
    'E  ;E',
    '(pre-compounded or Salt; Pepper)'
]

char3 = ["  ", "   ", "    ", "     ", "      ", " ,"]


def remove_1(string):
    for i in char1:
        string = string.replace(i, "")
    return string

def remove_2(string):
    for i in char2:
        string = string.replace(i, " ")
    return string

def remove_3(string):
    for i in char3:
        string = string.replace(i, "")
    return string
```

```python
def azureml_main(dataframe1 = None):   
    con = create_engine('mssql+pymssql://sqldbadmin:rYxz&5m~VjQrr&tm@mssql-gst-dev-ussc-01.database.windows.net/mssqldb-gst-dev-ussc-01')

    dataframe1 = []

## ----------------------------------------------------------------------------------------------

    df = pd.read_sql_query('select * from SFTRaw.Nsf',con=con)
    df.columns = df.columns.str.lower()
    #df = df[df.doc_type == "ARF"]
    df['cert'] = np.where(df['standard'] == '042', 'nsf 42', '')
    df = df[df['cert'] == 'nsf 42']
    df['cert'] = df[['cert']].values.tolist()
    df['grade'] = df['primary_trade_name']
    dataframe1.append(df[['grade','cert']])

## ----------------------------------------------------------------------------------------------

    # ingest and convert  nsf51 data
    df = pd.read_sql_query('select * from SFTRaw.Nsf',con=con)
    df.columns = df.columns.str.lower()
    #df = df[df.doc_type == "ARF"]
    df['cert'] = np.where(df['standard'] == '051', 'nsf 51', '')
    df = df[df['cert'] == 'nsf 51']
    df['cert'] = df[['cert']].values.tolist()
    df['grade'] = df['primary_trade_name']
    dataframe1.append(df[['grade','cert']])

## ----------------------------------------------------------------------------------------------

    # ingest and convert  nsf61 data
    df = pd.read_sql_query('select * from SFTRaw.Nsf',con=con)
    df.columns = df.columns.str.lower()
    #df = df[df.doc_type == "ARF"]
    df['cert'] = np.where(df['standard'] == '061', 'nsf 61', '')
    df = df[df['cert'] == 'nsf 61']
    df['cert'] = df[['cert']].values.tolist()
    df['grade'] = df['primary_trade_name']
    dataframe1.append(df[['grade','cert']])

## ----------------------------------------------------------------------------------------------

# ingest and convert  nsf61 23C and NSF61 82C
    df = pd.read_sql_query('select * from SFTRaw.Nsf',con=con)
    df.columns = df.columns.str.lower()
    print(df.columns)
    df = df[['primary_trade_name', 'nsf61_23c', 'nsf61_82c']]
    df = df.astype(str)
    # Transform nsf61
    df['nsf61_23c'] = np.where(df['nsf61_23c'] == 'Yes', 'nsf61 23c', '')
    df['nsf61_82c'] = np.where(df['nsf61_82c'] == 'Yes', 'nsf61 82c', '')
    #Transform
    df['nsf61_23c'] = df['nsf61_23c'].apply(lambda x: transform(x))
    df['nsf61_82c'] = df['nsf61_82c'].apply(lambda x: transform(x))
    df['cert'] = df['nsf61_82c'] + df['nsf61_23c']
    df['cert'] = df['cert'].apply(lambda x: list(filter(None, x)))
    df['grade'] = df['primary_trade_name']
    dataframe1.append(df[['grade','cert']])

## ----------------------------------------------------------------------------------------------

    # ingest and convert railway data
    df = pd.read_sql_query('select * from SFTRaw.Railway',con=con)
    df.columns = df.columns.str.lower()
    df['grade'] = df['material_name']
    df['cert'] = df['railway_certification']
    ## df['c1'] = df['norm'] + " " + df['specification'] + " " + df['classification']
    ## df['c2'] = df['norm'] + " " + df['specification']
    ## df['c3'] = df['specification'] + " " + df['classification']
    df['cert']= df[['cert']].values.tolist()
    df['cert'] = df['cert'].apply(lambda x: list(filter(None, x)))	
    dataframe1.append(df[['grade','cert']])

## ----------------------------------------------------------------------------------------------

    df = pd.read_sql_query('select * from SFTRaw.WaterApprovals',con=con)
    df.columns = df.columns.str.lower()
    print(df.columns)
    cols = df.columns[4:13]
    df[cols] = df[cols].apply(pd.to_datetime, errors='coerce')
    cols = list(df.select_dtypes(include=['datetime64']))
    df[cols] = df[cols].apply(lambda x: pd.to_datetime(x).dt.strftime('%m/%Y'))
    df[cols] = df[cols].apply(pd.to_datetime, errors='coerce')
    df = df.rename(columns={'ktw_bwgl_en16421_p2_23c': 'ktw_bwgl_23c', 'ktw_bwgl_en16421_p2_60c':'ktw_bwgl_60c', 'ktw_bwgl_en16421_p2_85c':'ktw_bwgl_85c', 'ktw_bwgl_en16421_p2_en16422':'en16421'})
    
        
    #getting Today's date
    Today=str(date.today())
    df['Today_date']=Today
    df['Today_date']=pd.to_datetime(df['Today_date'],errors='coerce')
    #Calling function for year and month extraction
    df=extract_year_month(df)
        
    #Filter 'wras' based on current year and month
    df['wras_23c']=np.where(((df['wras_23c_year']>df['Today_date_year'])|((df['wras_23c_month']>=df['Today_date_month'])&(df['wras_23c_year']==df['Today_date_year']))),df['wras_23c'].apply(str),"NaN")
    df['wras_60c']=np.where(((df['wras_60c_year']>df['Today_date_year'])|((df['wras_60c_month']>=df['Today_date_month'])&(df['wras_60c_year']==df['Today_date_year']))),df['wras_60c'].apply(str),"NaN")
    df['wras_85c']=np.where(((df['wras_85c_year']>df['Today_date_year'])|((df['wras_85c_month']>=df['Today_date_month'])&(df['wras_85c_year']==df['Today_date_year']))),df['wras_85c'].apply(str),"NaN")
    
     #Filter 'ktw' based on current year and month
    df['ktw_bwgl_23c']=np.where(((df['ktw_bwgl_23c_year']>df['Today_date_year'])|((df['ktw_bwgl_23c_month']>=df['Today_date_month'])&(df['ktw_bwgl_23c_year']==df['Today_date_year']))),df['ktw_bwgl_23c'].apply(str),"NaN")
    df['ktw_bwgl_60c']=np.where(((df['ktw_bwgl_60c_year']>df['Today_date_year'])|((df['ktw_bwgl_60c_month']>=df['Today_date_month'])&(df['ktw_bwgl_60c_year']==df['Today_date_year']))),df['ktw_bwgl_60c'].apply(str),"NaN")
    df['ktw_bwgl_85c']=np.where(((df['ktw_bwgl_85c_year']>df['Today_date_year'])|((df['ktw_bwgl_85c_month']>=df['Today_date_month'])&(df['ktw_bwgl_85c_year']==df['Today_date_year']))),df['ktw_bwgl_85c'].apply(str),"NaN")
    df['en16421']=np.where(((df['en16421_year']>df['Today_date_year'])|((df['en16421_month']>=df['Today_date_month'])&(df['en16421_year']==df['Today_date_year']))),df['en16421'].apply(str),"NaN")
    
    #Filter 'acs' based on current year and month
    df['acs_23c']=np.where(((df['acs_23c_year']>df['Today_date_year'])|((df['acs_23c_month']>=df['Today_date_month'])&(df['acs_23c_year']==df['Today_date_year']))),df['acs_23c'].apply(str),"NaN")
    df['acs_60c']=np.where(((df['acs_60c_year']>df['Today_date_year'])|((df['acs_60c_month']>=df['Today_date_month'])&(df['acs_60c_year']==df['Today_date_year']))),df['acs_60c'].apply(str),"NaN")
    df['acs_85c']=np.where(((df['acs_85c_year']>df['Today_date_year'])|((df['acs_85c_month']>=df['Today_date_month'])&(df['acs_85c_year']==df['Today_date_year']))),df['acs_85c'].apply(str),"NaN")

    df = df[[
    'grade',
    # 'nsf61_23c',
    # 'nsf61_82c',
    'wras_23c',
    'wras_60c',
    'wras_85c',
    'ktw_bwgl_23c',
    'ktw_bwgl_60c',
    'ktw_bwgl_85c',
    'en16421',
    'acs_23c',
    'acs_60c',
    'acs_85c']]

    col = df.columns[1:11]
    df[col] = df[col].apply(pd.to_datetime, errors='coerce')
    col = list(df.select_dtypes(include=['datetime64']))
    df[col] = df[col].apply(lambda x: pd.to_datetime(x).dt.strftime('%m/%Y'))
    df = df.astype(str)
        
    # # Transform nsf61
    # df['nsf61_23c'] = np.where(df['nsf61_23c'] == 'Yes', 'nsf61 23C', '')
    # df['nsf61_82c'] = np.where(df['nsf61_82c'] == 'Yes', 'nsf61 82C', '')

    # Transform wras
    df['wras_23c'] = df['wras_23c'].apply(lambda x: is_date(x))
    df['wras_60c'] = df['wras_60c'].apply(lambda x: is_date(x))
    df['wras_85c'] = df['wras_85c'].apply(lambda x: is_date(x))

    df['wras_23c'] = np.where(df['wras_23c'] == True, 'wras 23', "")
    df['wras_60c'] = np.where(df['wras_60c'] == True, 'wras 60', "")
    df['wras_85c'] = np.where(df['wras_85c'] == True, 'wras 85', "")


    # Transform acs
    df['acs_23c'] = df['acs_23c'].apply(lambda x: is_date(x))
    df['acs_60c'] = df['acs_60c'].apply(lambda x: is_date(x))
    df['acs_85c'] = df['acs_85c'].apply(lambda x: is_date(x))

    df['acs_23c'] = np.where(df['acs_23c'] == True, 'acs 23', "")
    df['acs_60c'] = np.where(df['acs_60c'] == True, 'acs 60', "")
    df['acs_85c'] = np.where(df['acs_85c'] == True, 'acs 85', "")


    # Transform ktw
    df['ktw_bwgl_23c'] = df['ktw_bwgl_23c'].apply(lambda x: is_date(x))
    df['ktw_bwgl_60c'] = df['ktw_bwgl_60c'].apply(lambda x: is_date(x))
    df['ktw_bwgl_85c'] = df['ktw_bwgl_85c'].apply(lambda x: is_date(x))
    df['en16421'] = df['en16421'].apply(lambda x: is_date(x))

    df['ktw_bwgl_23c'] = np.where(df['ktw_bwgl_23c'] == True, 'ktw bwgl en16421 P2 23C', "")
    df['ktw_bwgl_60c'] = np.where(df['ktw_bwgl_60c'] == True, 'ktw bwgl en16421 P2 60C', "")
    df['ktw_bwgl_85c'] = np.where(df['ktw_bwgl_85c'] == True, 'ktw bwgl en16421 P2 85C', "")
    df['en16421'] = np.where(df['en16421'] == True, 'ktw bwgl en16421 P2 en16422', "")

    # Transform
    # df['nsf61_23c'] = df['nsf61_23c'].apply(lambda x: transform(x))
    # df['nsf61_82c'] = df['nsf61_82c'].apply(lambda x: transform(x))
    df['wras_23c'] = df['wras_23c'].apply(lambda x: transform(x))
    df['wras_60c'] = df['wras_60c'].apply(lambda x: transform(x))
    df['wras_85c'] = df['wras_85c'].apply(lambda x: transform(x))
    df['acs_23c'] = df['acs_23c'].apply(lambda x: transform(x))
    df['acs_60c'] = df['acs_60c'].apply(lambda x: transform(x))
    df['acs_85c'] = df['acs_85c'].apply(lambda x: transform(x))
    df['ktw_bwgl_23c'] = df['ktw_bwgl_23c'].apply(lambda x: transform(x))
    df['ktw_bwgl_60c'] = df['ktw_bwgl_60c'].apply(lambda x: transform(x))
    df['ktw_bwgl_85c'] = df['ktw_bwgl_85c'].apply(lambda x: transform(x))
    df['en16421'] = df['en16421'].apply(lambda x: transform(x))

    # Create df['cert']
    df['cert'] = df['wras_23c'] + df['wras_60c'] + df['wras_85c'] + df['acs_23c'] + df['acs_60c'] + df['acs_85c'] + df['ktw_bwgl_23c'] + df['ktw_bwgl_60c'] + df['ktw_bwgl_85c'] + df['en16421']
    df['cert'] = df['cert'].apply(lambda x: list(filter(None, x)))
    dataframe1.append(df[['grade','cert']])

## ----------------------------------------------------------------------------------------------

    df = pd.read_sql_query("select * from SFTRaw.Feature where property_name='Approval'",con=con)

    df['VALUE_ASSMNT_SI'] = df['VALUE_ASSMNT_SI'].apply(lambda x: remove_1(x))
    df['VALUE_ASSMNT_SI'] = df['VALUE_ASSMNT_SI'].apply(lambda x: remove_2(x))
    df['VALUE_ASSMNT_SI'] = df['VALUE_ASSMNT_SI'].apply(lambda x: remove_3(x))
    df['VALUE_ASSMNT_SI'] = df['VALUE_ASSMNT_SI'].apply(lambda x: x.replace('(pre-compounded or Salt; Pepper)', ""))
    df['VALUE_ASSMNT_SI'] = df['VALUE_ASSMNT_SI'].apply(lambda x: x.replace(' 100% Color match', ""))
    df['VALUE_ASSMNT_SI'] = df['VALUE_ASSMNT_SI'].apply(lambda x: x.replace('COLOR MATCH', ""))
    df['VALUE_ASSMNT_SI'] = df['VALUE_ASSMNT_SI'].apply(lambda x: x.rstrip())
    df['VALUE_ASSMNT_SI'] = df['VALUE_ASSMNT_SI'].apply(lambda x: x.strip())
    df['VALUE_ASSMNT_SI'] = df['VALUE_ASSMNT_SI'].apply(lambda x: x.replace('Â', ""))

    df['TEMP_C2'] = df['TEMP_C2'].apply(lambda x: x.replace('Â', ""))
    df['TEMP_C2'] = df['TEMP_C2'].apply(lambda x: x.replace('\xa0', ""))
    df['TEMP_C2'] = df['TEMP_C2'].apply(lambda x: x.replace('Chyrsler', 'Chrysler'))
    df['TEMP_C2'] = df['TEMP_C2'].apply(lambda x: x.replace('Diamler', 'Daimler'))
    df['TEMP_C2'] = df['TEMP_C2'].apply(lambda x: x.rstrip())
    df['TEMP_C2'] = df['TEMP_C2'].apply(lambda x: x.strip())
    df['TEMP_C2'] = df['TEMP_C2'].apply(lambda x: x.replace('Stellantis PSA', 'Stellantis - PSA'))
    df['TEMP_C2']=np.where(((df['TEMP_C2']=='Mercedes-Benz Group')|(df['TEMP_C2']=='Mercedes-Benz Group (Daimler')|(df['TEMP_C2']=='Mercedes-Benz Group (Diamler)')),'Mercedes-Benz Group (Daimler)',df['TEMP_C2'])

    #Create grade & cert columns
    df['grade'] = df['PRODUCT_CD']
    df['cert'] = df[['TEMP_C2', 'VALUE_ASSMNT_SI']].values.tolist()
    df['cert'] = df['cert'].apply(lambda x: list(filter(None, x)))
    dataframe1.append(df[['grade','cert']])

    dff = pd.concat(dataframe1)

    x = dff.groupby('grade')

    a1, b1 = [], []
    for i in x:
        g = i[0]
        aa = i[1]['cert'].values
        #print(aa)
        if aa.any() != None:
            #print(aa)
            val = set(itertools.chain.from_iterable(aa))
            print(val)
            a1.append(g)
            b1.append(val)
        else:
            a1.append(g)
            b1.append(None)
            
    df_cert = pd.DataFrame({'grade':a1, 'certifications':b1})
    df_cert['certifications'] = df_cert['certifications'].apply(list)
    df_cert = df_cert.astype(str)
    dataframe1 = df_cert

    dataframe1.to_sql('df_cert_remapped_pipeline', con=con, schema='SftTransform', index=False, if_exists='replace')
    

    return dataframe1
```
