```python
import pandas as pd
import json
import ast
```

```python
df = pd.read_excel('data/Training_Data_27_01_25/Training_Data_27_01_25.xlsx')
```

```python
sl_nos = [str(i) + ":::" for i in list(df['Ref No.'])]
sl_nos
```

```python
outputs = [ast.literal_eval(i) for i in list(df['Output'])]
outputs[:10]
```

```python
def add_to_all_values(key, value, all_values, parent_key=None):    
    """    
    Adds the value to the corresponding list in all_values based on the key,    
    including handling nested structures by recursion.    
    Additionally, considers the parent_key to handle special cases.    
    """    
    # Special handling for context-specific keys  
    if key == 'standard' and parent_key:  
        if parent_key == 'RAILWAY_CERT':  
            key = 'railway_standard'  
        elif parent_key == 'WATER_CERT':  
            key = 'water_standard'  
  
    if key not in all_values:    
        all_values[key] = []    
  
    if isinstance(value, (str, int, float, dict)):    
        all_values[key].append(str(value))    
  
    if isinstance(value, list):    
        for item in value:    
            if isinstance(item, (dict, list)):    
                add_to_all_values(key, item, all_values, parent_key=key)  # Pass the current key as parent_key  
            else:    
                all_values[key].append(str(item))    
    elif isinstance(value, dict):    
        for sub_key, sub_value in value.items():    
            add_to_all_values(sub_key, sub_value, all_values, parent_key=key)  # Pass the current key as parent_key  
  
    return all_values

def capture_all_values(sample_dicts):  
    """  
    Traverses each dictionary in sample_dicts to populate an all_values dictionary  
    with values from both top-level and nested keys. Returns the all_values dictionary.  
    """  
    all_values = {  
        # Initialize the structure here with the keys you expect to encounter  
        'GRADE': [], 'APPLICATION': [], 'BRAND': [], 'POLYMER': [], 'PROPERTY': [],  
        'FILLER': [], 'FEATURE': [], 'PROCESSING': [], 'DELIVERY_FORM': [],  
        'COMPETITOR_GRADE': [], 'AUTO_CERT': [], 'RAILWAY_CERT': [], 'WATER_CERT': [],  
        'NSF_CERT': [], 'property_name': [], 'modifier': [], 'property_type': [],  
        'filler_name': [], 'total_load': [], 'oem': [], 'certs': [],  
        'railway_standard': [], 'hazard_level': [], 'req_set': [], 'water_standard': [],  
        'temp': [], 'INDUSTRY': [], 'REGION': []  
    }  
  
    for sample_dict in sample_dicts:  
        for key, value in sample_dict.items():  
            all_values = add_to_all_values(key, value, all_values)  
  
    for key in all_values:  
        # Convert list to set to remove duplicates, then back to list  
        all_values[key] = list(set(all_values[key])) 
  
    return all_values
```

```python
all_labelled_unique_values = capture_all_values(outputs)  
```

```python
with open("data/Training_Data_27_01_25/all_labelled_unique_values_27_01_25.json", 'w', encoding='utf-8') as fp:
    json.dump(all_labelled_unique_values, fp, ensure_ascii=False, indent=3)
```

```python
def add_to_all_values2(key, value, all_values, prefix="", parent_key=None):      
    """      
    Adds the value with a prefix to the corresponding list in all_values based on the key,      
    including handling nested structures by recursion.      
    Additionally, considers the parent_key to handle special cases.      
    """      
    # Special handling for context-specific keys    
    if key == 'standard' and parent_key:    
        if parent_key == 'RAILWAY_CERT':    
            key = 'railway_standard'    
        elif parent_key == 'WATER_CERT':    
            key = 'water_standard'    
    
    if key not in all_values:      
        all_values[key] = []      
    
    if isinstance(value, (str, int, float, dict)) and value:      
        all_values[key].append(prefix + str(value))  # Prepend prefix here  
      
    if isinstance(value, list):      
        for item in value:      
            if isinstance(item, (dict, list)):      
                add_to_all_values2(key, item, all_values, prefix, parent_key=key)  # Pass the current key as parent_key    
            elif item:      
                all_values[key].append(prefix + str(item))  # Prepend prefix here    
    elif isinstance(value, dict):      
        for sub_key, sub_value in value.items():      
            add_to_all_values2(sub_key, sub_value, all_values, prefix, parent_key=key)  # Pass the current key as parent_key    
    
    return all_values  
  
def capture_all_values2(sample_dicts, prefixes):    
    """    
    Traverses each dictionary in sample_dicts to populate an all_values dictionary    
    with values from both top-level and nested keys, including a prefix for each value.    
    Returns the all_values dictionary.    
    """    
    all_values = {    
        # Initialize the structure here with the keys you expect to encounter    
        'GRADE': [], 'APPLICATION': [], 'BRAND': [], 'POLYMER': [], 'PROPERTY': [],    
        'FILLER': [], 'FEATURE': [], 'PROCESSING': [], 'DELIVERY_FORM': [],    
        'COMPETITOR_GRADE': [], 'AUTO_CERT': [], 'RAILWAY_CERT': [], 'WATER_CERT': [],    
        'NSF_CERT': [], 'property_name': [], 'modifier': [], 'property_type': [],    
        'filler_name': [], 'total_load': [], 'oem': [], 'certs': [],    
        'railway_standard': [], 'hazard_level': [], 'req_set': [], 'water_standard': [],    
        'temp': [], 'INDUSTRY': [], 'REGION': []  
    }    
    
    for sample_dict, prefix in zip(sample_dicts, prefixes):  # Use zip to iterate over both lists simultaneously  
        for key, value in sample_dict.items():    
            all_values = add_to_all_values2(key, value, all_values, prefix=prefix)  # Pass prefix here  
    
    for key in all_values:    
        # Convert list to set to remove duplicates, then back to list    
        all_values[key] = list(set(all_values[key]))
        all_values[key] = [x for x in all_values[key] if x]
    
    return all_values  
```
